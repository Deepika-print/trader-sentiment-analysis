```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load datasets
trades = pd.read_csv("../data/historical_data.csv")
sentiment = pd.read_csv("../data/fear_greed_index.csv")

print("Data loaded successfully")
```

Here I merge trading data with market sentiment using date.

```python
# Convert date columns
trades['time'] = pd.to_datetime(trades['time'], errors='coerce')
sentiment['Date'] = pd.to_datetime(sentiment['Date'], errors='coerce')

# Create common date column
trades['date'] = trades['time'].dt.date
sentiment['date'] = sentiment['Date'].dt.date

# Merge datasets
data = pd.merge(trades, sentiment, on='date', how='left')

print("Merge done")
```

Now I check profit, win rate and trading behavior.

```python
# Clean data
data['closedPnL'] = pd.to_numeric(data['closedPnL'], errors='coerce')
data = data.dropna(subset=['Classification'])

# Create profit column
data['profit'] = data['closedPnL'] > 0

print("Data cleaned")
```

```python
# Average profit
print("\nAverage Profit:")
print(data.groupby('Classification')['closedPnL'].mean())
```

```python
# Win rate
print("\nWin Rate:")
print(data.groupby('Classification')['profit'].mean())
```

```python
# Leverage usage
print("\nAverage Leverage:")
print(data.groupby('Classification')['leverage'].mean())
```

```python
print("\n--- Insight ---")
print("During Greed, traders take more risk and profits are higher but less stable.")
print("During Fear, trading is more consistent with better win rate.")
```

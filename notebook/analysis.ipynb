```python
import pandas as pd
```

```python
trades = pd.read_csv("../data/historical_data.csv")
sentiment = pd.read_csv("../data/fear_greed_index.csv")
```

Here I match trades with market sentiment using date.

```python
trades['time'] = pd.to_datetime(trades['time'])
sentiment['Date'] = pd.to_datetime(sentiment['Date'])

trades['date'] = trades['time'].dt.date
sentiment['date'] = sentiment['Date'].dt.date

data = pd.merge(trades, sentiment, on='date')
```

Now I check profit and win rate.

```python
data['closedPnL'] = pd.to_numeric(data['closedPnL'], errors='coerce')
data['profit'] = data['closedPnL'] > 0

data.groupby('Classification')['closedPnL'].mean()
```
